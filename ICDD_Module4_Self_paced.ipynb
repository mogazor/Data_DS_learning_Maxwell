### 🧹 Data Cleaning Step 1: Rename Columns

```python
# Let's make our column names more readable and consistent
column_rename = {
    'Country.of.Origin': 'origin_country',
    'Farm.Name': 'farm_name',
    'Lot.Number': 'lot_number',
    'Mill': 'mill',
    'ICO.Number': 'ico_number',
    'Company': 'company',
    'Altitude': 'altitude',
    'Region': 'region',
    'Producer': 'producer',
    'Number.of.Bags': 'bag_count',
    'Bag.Weight': 'bag_weight',
    'Processing.Method': 'processing_method',
    'Flavor': 'flavor_score',
    'Aftertaste': 'aftertaste_score',
    'Acidity': 'acidity_score',
    'Body': 'body_score',
    'Balance': 'balance_score',
    'Uniformity': 'uniformity_score',
    'Clean.Cup': 'clean_cup_score',
    'Sweetness': 'sweetness_score',
    'Total.Cup.Points': 'total_score'
}

coffee_clean = coffee_data.rename(columns=column_rename)
```

```python
# Let's check for missing values
print("\n🔍 Missing values in each column:")
missing_values = coffee_clean.isnull().sum()
display(missing_values[missing_values > 0])
```

### 💡 Understanding Missing Values:
Missing values can occur for many reasons:
*   Data wasn't collected
*   Sensor malfunctions
*   Human error in data entry
*   Information wasn't available

### 🧹 Data Cleaning Step 2: Handling Missing Values
Let's make informed decisions about how to handle missing values

```python
# First, let's look at what's in our altitude column
print("Altitude column info:")
print(coffee_clean['altitude'].dtype)
print("\nSample of altitude values:")
print(coffee_clean['altitude'].head(10))

# Check if there are any non-numeric values
print("\nUnique values in altitude column:")
print(coffee_clean['altitude'].unique()[:10])
```

```python
# Clean altitude column
def clean_altitude(alt):
    if isinstance(alt, str):
        # Extract numbers from strings like "1200m" or "1200-1500"
        import re
        numbers = re.findall(r'\d+', str(alt))
        if numbers:
            # If range is given, take the average
            return sum(int(num) for num in numbers) / len(numbers)
    elif pd.notnull(alt):
        return float(alt)
    return np.nan

# Clean the altitude column
coffee_clean['altitude'] = coffee_clean['altitude'].apply(clean_altitude)

# Now we can fill missing values with median
coffee_clean['altitude'] = coffee_clean['altitude'].fillna(coffee_clean['altitude'].median())

# Verify the results
print("\nAltitude statistics after cleaning:")
print(coffee_clean['altitude'].describe())
```

```python
# For text columns, we'll fill with 'Unknown'
text_columns = ['farm_name', 'mill', 'region', 'producer']
for col in text_columns:
    coffee_clean[col] = coffee_clean[col].fillna('Unknown')
```

### 🧹 Data Cleaning Step 3: Detecting Outliers

```python
print("\n🔍 Let's look for outliers in our numerical scores")
```

```python
# Create a box plot for all scoring categories
plt.figure(figsize=(15, 6))
score_columns = [col for col in coffee_clean.columns if col.endswith('_score')]
coffee_clean[score_columns].boxplot()
plt.xticks(rotation=45)
plt.title('Distribution of Coffee Scores')
plt.ylabel('Score')
plt.tight_layout()
plt.show()
```

```python
# Calculate outlier boundaries for total score
Q1 = coffee_clean['total_score'].quantile(0.25)
Q3 = coffee_clean['total_score'].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
```

```python
print(f"\n📊 Total Score Statistics:")
print(f"Lower bound for outliers: {lower_bound:.2f}")
print(f"Upper bound for outliers: {upper_bound:.2f}")
print(f"Number of outliers below: {len(coffee_clean[coffee_clean['total_score'] < lower_bound])}")
print(f"Number of outliers above: {len(coffee_clean[coffee_clean['total_score'] > upper_bound])}")
```

```python
# 📊 Exploratory Data Analysis (EDA)
print("\n🔍 Let's explore some patterns in our data!")
```

```python
# 1. Distribution of Total Scores
plt.figure(figsize=(12, 6))
sns.histplot(data=coffee_clean, x='total_score', bins=30)
plt.title('Distribution of Total Coffee Scores')
plt.xlabel('Total Score')
plt.ylabel('Count')
plt.show()
```

```python
# Calculate some basic statistics
print("\n📊 Total Score Statistics:")
print(coffee_clean['total_score'].describe())
```

```python
# 2. Average Scores by Country (top 10 countries)
plt.figure(figsize=(12, 6))
country_avg = coffee_clean.groupby('origin_country')['total_score'].mean().sort_values(ascending=False).head(10)
sns.barplot(x=country_avg.index, y=country_avg.values)
plt.title('Average Coffee Scores by Country (Top 10)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# 3. Correlation between different scores
plt.figure(figsize=(12, 10))
score_correlation = coffee_clean[score_columns].corr()
sns.heatmap(score_correlation, annot=True, cmap='coolwarm', center=0)
plt.title('Correlation between Different Score Components')
plt.tight_layout()
plt.show()
```

```python
# 4. Relationship between Altitude and Score
plt.figure(figsize=(10, 6))
sns.scatterplot(data=coffee_clean, x='altitude', y='total_score', alpha=0.5)
plt.title('Relationship between Altitude and Coffee Score')
plt.xlabel('Altitude')
plt.ylabel('Total Score')
plt.show()
```
